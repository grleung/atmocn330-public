# Part 1: The Atmosphere (10pts)

For this section, we'll be working with radiosonde data. Every day, hundreds of radiosondes are launched all over the world! Radiosondes measure temperature, pressure, humidity, U/V windspeed, and the latitude/longitude/altitude position. These coordinated launches (usually at 0Z and 12Z) are assimilated into operational forecast systems, and are a very important way that we can get information about the actual thermodynamic state of our atmosphere. 

You can see a full map of sounding locations and download soundings from the University of Wyoming archive: https://weather.uwyo.edu/upperair/sounding.shtml.

## 1.2. Plot atmospheric sounding with different vertical coordinates

Now, we want to plot the vertical profile of temperature for the above sounding (in the variable `df`). Here's a simple example of how to plot the temperature vs. height:

```python
# set up a figure with just one sub-plot axis
fig, ax = plt.subplots(1, 1)

df.plot(
    ax=ax,  # what axis to plot on
    x="temperature",  # what to plot on the x-axis, this is temperature in degrees Celsius
    y="height",  # what to plot on the y-axis, this is altitude above mean sea level in meters,
    label="Davenport",  # label the location of our sounding
    grid=True,  # draw grid lines to make the plot easier to read
)


# always label your axes!
ax.set_xlabel("Temperature (degC)")
ax.set_ylabel("Altitude (m a.s.l.)")  # a.s.l. = above sea level
```

# Part 2: The Ocean (10pts)

Similar to how the atmosphere has a global radiosonde network, the thermodynamic state of the ocean is also measured with a network of ocean buoys called Argo floats. Argo floats measure the temperature and salinity of saltwater (some floats also measure properties to describe oceanic biology and chemistry). Unlike radiosondes, which are released at coordinated times across the globe, ARGO floats are launched from ships, drift with the ocean, and descend and ascend on a regular timer to sample vertical profiles of the ocean.

You can read more about the Argo network and see a map of where the floats are here: https://argo.ucsd.edu.
