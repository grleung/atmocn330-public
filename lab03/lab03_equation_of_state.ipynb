# Part 1: Atmosphere (10 pts)

Recall that the equation of state for the atmosphere is given by

$$ p = \rho R_d T_v$$

where $R_d=287\ J\ K^{-1}\ kg^{-1}$ is the specific gas constant for dry air and $T_v\approx(1+0.61q)T$ is the virtual temperature.

It's good practice to define any constants you may use repeatedly at the top of your function rather than hard-coding the numbers, so that you only need to change it in one place! 

For this lab, I've defined useful constants below, so you can now use `Rd`, `Rv`, and `eps` in the rest of the notebook without having to re-define them. If you use them in a different notebook (or use other constants I didn't define here), you'll need to define your constants again.

Remember to label your constants with a comment (beginning with a # sign), including the units.

**Note:** I typically use the writing convention of defining units using square brackets []. You can use whatever convention you want, but stick to something consistent to make it easier to scan.

```python
Rd = 287  # specific gas constant for dry air [J K^-1 kg^-1]
Rv = 461.5  # specific gas constant for water vapor [J K^-1 kg^-1]
eps = (
    Rd / Rv
)  # ratio of gas constant for dry air to gas constant for water vapor [unitless]
```

## 1.1 Intro to functions

**Goal:** Use the ideal gas law to set up a function to return pressure for a given density, temperature, and specific humidity.

A *function* is a way to encapsulate code so that you don't need to repeat calculations over and over again. A function takes some *inputs* (sometimes called parameters or arguments), and returns some *output(s)*. 

For example:

```python
def calculate_pressure(rho, T, q):
    # Calculate pressure (p) [hPa] as a function of density (rho) [kg m^-3], temperature (T) [K], and specific humidity (q) [kg kg^-1]

    p = (
        rho * Rd * (1 + ((1 - (1 / eps)) * q)) * T
    )  # notice that this is the ideal gas law we wrote above, but in code form; be careful about parentheses!

    # convert pascals to hectopascals
    p = p / 100

    return p
```

Always comment your code! At minimum, you should define what the input(s) and output(s) of your function are, including what units the function expects.

Now we can use our function to make sure it works.

**Important:** Notice that the function returns the value of pressure in *hPa*, not *Pa*. Be careful with units!

```python
p = calculate_pressure(
    1.225,  # density in kg/m3
    288.2,  # temp in K
    0.0000,  # q in kg/kg
)

print(p)
```

Woah that is a lot of decimal places, plus no units! Computers carry full floating-point precision. Let's round to 1 decimal place to be more readable + make sure to specify units in our output.

```python
print(round(p, 1), "hPa")
```

## 1.3 Plotting density as a 1D function of other state variables

Now that we have a function, we can very quickly calculate density for lots of values (hooray!). This is very useful if we want to visualize how density varies as a function of the other state variables, for example temperature.

The following code block shows how to calculate density for an array of temperatures, assuming a fixed p=1000 hPa and q=0 kg/kg. 

```python
# this generates an array of temperature values ranging from 200 to 320, with increments of 10
T_array = np.arange(200, 330, 10)  # temperatures [K]

print("Temperatures:")
print(T_array, "K")

# we can pass the whole array of temperatures into our calculate_density function to calculate many densities at once
rho_array = calculate_density(1000, T_array, 0)

print("\nDensities:")
print(rho_array, "kg/m3")
```

```python
fig, ax = plt.subplots(1, 1)

ax.plot(
    T_array,  # temperature on the x-axis
    rho_array,  # density on the y-axis
)

# always label your axes!
ax.set_xlabel("Temperature (K)")
ax.set_ylabel("Density (kg/m3)")

# set a grid to make the plot easier to read
ax.grid(True)
```

## 1.4 Plotting density as a 2D function of temperature and pressure

We can also explore how density varies as a function of both temperature AND pressure, by defining a meshgrid.

```python
T_array = np.arange(230, 320, 10)  # temperatures [K]
p_array = np.arange(100, 1001, 50)  # pressures [hPa]

T_mesh, p_mesh = np.meshgrid(
    T_array,
    p_array,
)

rho_mesh = calculate_density(p_mesh, T_mesh, 0)
```

```python
fig, ax = plt.subplots(1, 1)

c = ax.contourf(
    T_array,  # x-axis
    p_array,  # y-axis
    rho_mesh,  # values to be plotted in contours
)
fig.colorbar(c, ax=ax, label="Density (kg/m^3)")

# always label your axes!
ax.set_xlabel("Temperature (K)")
ax.set_ylabel("Pressure (hPa)")

# let's flip the pressure axis, so values close to the surface are near the bottom
ax.invert_yaxis()
```

For example:

```python
rho = gsw.rho(33, 5, 100)
print(round(rho, 1), "kg m^-3")
```
