## 3. Quick calculation and plot check

Let's do a quick calculation with MetPy and generate a simple line plot to make sure matplotlib is working. 

````python
# Simple dewpoint calculation
temp = 20 * units.degC
rh = 50 * units.percent
dewpoint = mpcalc.dewpoint_from_relative_humidity(temp, rh)

print(f"MetPy Check: Dewpoint is {dewpoint:.1f}\n")
````

````python
dates = pd.date_range("2026-09-01", periods=5)
temps = [68, 72, 75, 70, 65]  # Daily high temperatures (°F)

data = xr.DataArray(temps, coords=[dates], dims=["time"])
fig, ax = plt.subplots(figsize=(7, 4))

# TODO: Change the color of this line by adding the keyword color='black'
data.plot(ax=ax)

# TODO: Change this to give your plot a real title
ax.set_title("Customized Title")

# TODO: Change this to give your plot an axis label
ax.set_ylabel('axis label')

plt.show()
````
