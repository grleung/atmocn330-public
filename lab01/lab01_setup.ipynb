```python
# Import core packages using standard nicknames
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import xarray as xr
import metpy.calc as mpcalc

# We want to call a subset of metpy called units so we can attach physical units like degC or hPa to our data
from metpy.units import units
```

# 3. Quick calculation and plot check

Let's do a quick calculation with MetPy and generate a simple line plot to make sure matplotlib is working. 

```python
# Simple dewpoint calculation
temp = 20 * units.degC
rh = 50 * units.percent
dewpoint = mpcalc.dewpoint_from_relative_humidity(temp, rh)

print(f"For a temperature of {temp:.1f} and RH of {rh:.1f}, the dewpoint is {dewpoint:.1f}\n")
```

```python
# generating our xarray to plot
dates = pd.date_range("2026-08-12", periods=5)
temps = [77,79,81,84,81]  # Daily high temperatures (°F)
data = xr.DataArray(temps, coords=[dates], dims=["time"])

# set up the plot
fig, ax = plt.subplots(figsize=(7, 4))

data.plot(ax=ax)

# set our plot title
ax.set_title("Forecast daily high temperatures for Madison, WI")

#always label your axes!
ax.set_ylabel('Air Temperature (°F)')

plt.show()
```
