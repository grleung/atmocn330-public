Repeating a very important point from earlier: When a parcel is *subsaturated*, it follows the **dry adiabat**. But when a parcel is *saturated*, it follows the **moist adiabat**. 

The point when the parcel becomes saturated is called the **lifting condensation level** (LCL). To find the LCL, follow these steps:

1. Find the parcel's surface temperature and dewpoint.
2. Follow the dry adiabat up from the surface temperature.
3. Follow the mixing ratio line up from the surface dewpoint.
4. Find the intersection of the dry adiabat and mixing ratio line. 
5. Look along a horizontal line from that intersection to the y-axis labels to read the pressure value of the LCL. 
6. To read the LCL temperature, follow the isotherm diagonally from the intersection point in step 4.

# Part 2: (Simple) parcel lifting model (15 pts)

In this next section, we are going to explore the parcel's state as it is lifted from the surface until the LCL. To do this, we will use a simple parcel lifting model contained in the function `lift_parcel` below.

**You do not need to edit the code block below. The model should work out of the box!** Your job for this lab is just to change the input values and see what happens in order to answer the questions that follow.

A brief description of our parcel model:
1. We will lift a parcel from 1000 to 100 hPa in 1 hPa increments.
2. Below saturation:
    - parcel temperature `Temp` cools with the dry adiabatic lapse rate (using Poisson's equation)
    - mixing ratio `wv` is conserved
    - saturation mixing ratio `ws` is updated for the new value of `Temp` according to the Bolton form of Clausius-Clapeyron
3. Above saturation (checked by comparing `wv` to `ws`):
    - at first follows Step 2 (for dry lifting) to make an initial guess for `Temp` and `wv`, and see if the parcel is saturated (`wv>ws`)
    - if the parcel is saturated, any excess vapor above saturation (`wv-ws`) is condensed and converted to liquid water (saved as `wl`)
    - that condensation releases latent heat ($q = L_v \times (w_v-w_s)$)
    - latent heat release is used to adjust the temperature ($T = T_{initial\ guess} + (q\times c_p)$) and slow down rate of cooling from adiabatic expansion

```python
def lift_parcel(
    temp=20.0,
    dewpoint=None,
    pres_start=1000.0,
    pres_end=100.0,
    dp=1.0,
    allow_condensation=True,
):
    """Simple parcel lifting model assuming saturation adjustment (all supersaturation is immediately condensed as liquid water)

    Args:
        temp: Surface air temperature in degrees C. Defaults to 20.
        dewpoint: Surface dewpoint in degrees C. Defaults to None.
        pres_start: Surface pressure in hPa. Defaults to 1000.
        pres_end: Ending pressure in hPa. Defaults to 100.
        dp: Pressure level increment in hPa. Defaults to 1.
        allow_condensation: Flag for allowing condensation or not. If condensation is not allowed, parcel is lifted dry adiabatically only

    Returns:
        pd.DataFrame: pandas dataframe containing parcel trajectory; columns are 'p' (pressure; hPa), 'Temp' (air temp; degC), 'wv' (vapor mixing ratio: kg/kg), 'ws' (saturation vapor mixing ratio: kg/kg), 'rh' (relative humidity: %), 'wl' (liquid water mixing ratio: kg/kg), and 'latheat' (cumulative latent heat release in trajectory; J/kg)
    """

    # set up a dataframe to hold the output from the function
    if allow_condensation:
        trajectory = pd.DataFrame(
            columns=["p", "Temp", "wv", "ws", "rh", "wl", "latheat"]
        )
    else:
        trajectory = pd.DataFrame(columns=["p", "Temp", "wv", "ws", "rh"])

    # fill in pressure values at regular intervals from starting to ending pressure in hPa
    trajectory["p"] = np.arange(pres_start, pres_end - dp, -dp)

    # fill in surface temperature in degrees C
    trajectory.loc[0, "Temp"] = temp

    # fill in initial saturation mixing ratio calculated from surface temp/pressure
    trajectory.loc[0, "ws"] = calculate_w_from_e(
        calculate_es_from_temp(trajectory.loc[0, "Temp"]), trajectory.loc[0, "p"] * 100
    )

    if dewpoint:
        # calculate initial water vapor mixing ratio based on dewpoint
        trajectory.loc[0, "wv"] = calculate_w_from_e(
            calculate_es_from_temp(dewpoint), trajectory.loc[0, "p"] * 100
        )
    else:
        # if no dewpoint is given, assume parcel lifting is dry
        trajectory.loc[0, "wv"] = 0.0

    # initial RH
    trajectory.loc[0, "rh"] = 100 * trajectory.loc[0, "wv"] / trajectory.loc[0, "ws"]

    if allow_condensation:
        # fill liquid water column with all zero's by default
        trajectory["wl"] = 0.0
        trajectory["latheat"] = 0.0

    for i in range(1, len(trajectory)):
        # calculate new temperature assuming dry adiabatic lifting using Poisson's formula
        trajectory.loc[i, "Temp"] = (
            (trajectory.loc[i - 1, "Temp"] + 273.15)
            * (trajectory.loc[i, "p"] / trajectory.loc[i - 1, "p"]) ** kappa
        ) - 273.15

        # update saturation mixing ratio based on new temperature
        trajectory.loc[i, "ws"] = calculate_w_from_e(
            calculate_es_from_temp(trajectory.loc[i, "Temp"]),
            trajectory.loc[i, "p"] * 100,
        )

        # initial guess: conserve vapor mixing ratio, liquid water mixing ratio, and cumulative latent heat released from previous level
        trajectory.loc[i, "wv"] = trajectory.loc[i - 1, "wv"]

        if allow_condensation:
            trajectory.loc[i, "wl"] = trajectory.loc[i - 1, "wl"]
            trajectory.loc[i, "latheat"] = trajectory.loc[i - 1, "latheat"]

            # if the parcel is saturated, then update the initial guess based on latent heat release
            # check if supersaturated
            if trajectory.loc[i, "wv"] > trajectory.loc[i, "ws"]:

                # any excess water vapor above saturation is condensed (remove from water vapor and add to liquid)
                condensed_vapor = trajectory.loc[i, "wv"] - trajectory.loc[i, "ws"]
                trajectory.loc[i, "wl"] += condensed_vapor
                trajectory.loc[i, "wv"] = trajectory.loc[i, "ws"]

                # excess water from condensation warms temperature at this level
                trajectory.loc[i, "Temp"] += condensed_vapor * lv / cp

                # save latent heating for plotting
                trajectory.loc[i, "latheat"] += condensed_vapor * lv

        # calculate RH
        trajectory.loc[i, "rh"] = (
            100 * trajectory.loc[i, "wv"] / trajectory.loc[i, "ws"]
        )

    return trajectory.astype("float")
```

```python
def plot_parcel_traj(axes, traj, color=None, name=None):
    if "latheat" in traj.columns:
        ax_temp, ax_mr, ax_rh, ax_q = axes
    else:

        ax_temp, ax_mr, ax_rh = axes

    # plot temperature vs. pressure
    traj.plot(ax=ax_temp, x="Temp", y="p", grid=True, label=f"T {name}", color=color)

    # plot vapor mixing ratio vs. pressure
    traj.plot(ax=ax_mr, x="wv", y="p", grid=True, label=f"$w_v$ {name}", color=color)
    traj.plot(
        ax=ax_mr, x="ws", y="p", grid=True, label=f"$w_s$ {name}", ls="--", color=color
    )
    if "latheat" in traj.columns:
        traj.plot(
            ax=ax_mr,
            x="wl",
            y="p",
            grid=True,
            label=f"$w_l$ {name}",
            ls=":",
            color=color,
        )

    # plot rh vs. pressure
    traj.plot(ax=ax_rh, x="rh", y="p", grid=True, label=f"RH {name}", color=color)

    if "latheat" in traj.columns:
        traj.plot(
            ax=ax_q,
            x="latheat",
            y="p",
            grid=True,
            label=f"Total Heating {name}",
            color=color,
        )

    # flip pressure axis
    ax_temp.yaxis.set_inverted(True)

    # label axes
    ax_temp.set_ylabel("Parcel Pressure (hPa)")
    ax_temp.set_xlabel("Temperature (°C)")
    ax_mr.set_xlabel(f"Mixing Ratio (kg kg$^{{-1}}$)")
    ax_rh.set_xlabel("Relative Humidity (%)")
    if "latheat" in traj.columns:
        ax_q.set_xlabel("Heating (J kg$^{{-1}}$)")

    # add legend
    ax_mr.legend()

    # add marker for 100% RH
    ax_rh.axvline(100, ls=":", color="gray")

    return axes
```

Remember that **equivalent potential temperature** quantifies both a parcel's sensible heat (i.e., its temperature/potential temperature) AND its latent heat (the potential energy stored in the gas phase water vapor that could be condensed out). 

It turns out that the calculation you did in Question 12 (adding the latent heat released in condensing all the water vapor in the parcel directly to the parcel's temperature at 1000 hPa) *was* calculating $\theta_e$!

Let's go back to our Skew-T and think about equivalent potential temperature. To calculate $\theta_e$, do the following:

1. Find the parcel's LCL.
2. Below the LCL, trace the parcel up vertically along a dry adiabat.
3. Above the LCL, trace the parcel up vertically along a moist adiabat. 
4. Lift the parcel until the dry and moist adiabats are roughly parallel (e.g., up to 100 hPa or so)
5. Lower the parcel along a dry adiabat until you reach 1000 hPa. This is the **equivalent potential temperature**

```python
fig = plt.figure(figsize=(9, 9))

# initialize the Skew-T axis
skew = SkewT(fig, rotation=45)

T = 20
Td = 10
p = 1000

skew.ax.scatter(T, p, color="black")
skew.ax.scatter(Td, p, color="blue")

# axis labels
skew.ax.set_xlabel("Temperature (degC)")
skew.ax.set_ylabel("Pressure (hPa)")

# adding temperature labels at the top of our plot to help read values
skew.ax.tick_params(
    axis="x",
    top=True,
    labeltop=True,
    bottom=True,
    labelbottom=True,
)

# add dry adiabat
skew.plot_dry_adiabats(np.arange(-40, 50, 5) * units("degC"))

# add moist adiabat
skew.plot_moist_adiabats(np.arange(-40, 50, 5) * units("degC"))

# add mixing ratio lines
skew.plot_mixing_lines()

plt.show()
```
