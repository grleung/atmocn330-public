For this next portion, we will be using some actual atmospheric science data! 

You will be plotting a pressure level-latitude cross-section of the mean temperatures and potential temperatures for August 2026. The values are averaged zonally, i.e., across all longitudes. Your figure should end up looking similar to the figures we looked at in Lec 10.

These data are from ERA-5 (ECMWF ReAnalysis; ECMWF is the European met center). 

````python
# read the netCDF file (ending in .nc)
era_data = xr.open_dataset("data/august2026_era5_mean.nc")

# for simplicity, we'll extract the values and turn them into numpy arrays like we've been using in the rest of the class so far
t_K = era_data["temp_K"].values
p_hPa, latitudes = np.meshgrid(
    era_data["pressure_hPa"].values, era_data["latitude"].values, indexing="ij"
)
````

This gives us an array for `t_K` (temperatures in Kelvin), where each row is one pressure level and each column is one latitude. 

````python
t_K
````

Now, we'll plot temperature below. 

````python
fig, ax = plt.subplots(1, 1)

# make a contour plot
# ax.contourf takes arguments as: (x-values, y-values, color shading values)
c = ax.contourf(latitudes, p_hPa, t_K)

# make a colorbar
plt.colorbar(c, ax=ax, label="Temperature (K)")

# label axes
ax.set_xlabel("Latitude (degN/S)")
ax.set_ylabel("Pressure (hPa)")

# flip y-axis so surface pressure is near the bottom
ax.invert_yaxis()

plt.show()
````

# Part 3: Practice with Skew-Ts (13 pts)

Next, we are going to work with vertical profiles of temperature and learn how to plot them on a skew-T. 

Let's pull out one temperature profile from the equator.

````python
era_data = xr.open_dataset("data/august2026_era5_mean.nc").sel(latitude=0)

# for simplicity, we'll extract the values and turn them into numpy arrays like we've been using in the rest of the class so far
t_K = era_data["temp_K"].values
p_hPa = era_data["pressure_hPa"].values

# converting temperature to celsius for convenient plotting
t_C = t_K - 273.15
````

On to our plots! We could just plot temperature versus pressure (like we did in Lab01), but as we discussed in Lec07, a skewT-logP plot is much more convenient!

Many atmospheric scientists plot skewTs using a Python package called `metpy`, which is developed and maintained by NSF Unidata. Along with many other useful calculation tools, `metpy` has a nice set of functions for plotting skewTs.

````python
fig = plt.figure(figsize=(9, 9))

# initialize the Skew-T axis
skew = SkewT(fig, rotation=45)

# plot as normal, except we put `skew.` in front of the `ax.plot(...)`
skew.ax.plot(t_C, p_hPa)

# axis labels
skew.ax.set_xlabel("Temperature (degC)")
skew.ax.set_ylabel("Pressure (hPa)")

# adding temperature labels at the top of our plot to help read values
skew.ax.tick_params(
    axis="x",
    top=True,
    labeltop=True,
    bottom=True,
    labelbottom=True,
)

skew.ax.set_xlim(-20, 60)
skew.ax.set_ylim(1000)

plt.show()
````

One reason why SkewTs are so useful is that we can add more lines onto them to represent typical atmospheric motions. 

For example, the line ```skew.plot_dry_adiabats(...)``` below plots the path of several dry adiabatic parcels as red dashed lines.

````python
fig = plt.figure(figsize=(9, 9))

# initialize the Skew-T axis
skew = SkewT(fig, rotation=45)

# plot as normal, except we put `skew.` in front of the `ax.plot(...)`
skew.ax.plot(t_C, p_hPa)

# axis labels
skew.ax.set_xlabel("Temperature (degC)")
skew.ax.set_ylabel("Pressure (hPa)")

# adding temperature labels at the top of our plot to help read values
skew.ax.tick_params(
    axis="x",
    top=True,
    labeltop=True,
    bottom=True,
    labelbottom=True,
)
skew.ax.set_xlim(-20, 60)
skew.ax.set_ylim(1000)

# NEW: plotting dry adiabats
skew.plot_dry_adiabats(np.arange(-40, 100, 10) * units("degC"))

plt.show()
````
