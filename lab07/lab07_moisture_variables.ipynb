# Part 4: Introducing dewpoint temperature (4 pts)

It is often convenient to express moisture in temperature units, so we can compare more directly to the ambient air temperature.

The most common such variable is the **dewpoint temperature** ($T_d$), which is the temperature to which air must be cooled at constant pressure and water vapor content (quantified using either $q$, $w$, or $e$) to reach saturation.

Mathematically, we can find the dewpoint by inverting the C-C relationship to solve for the temperature $T_d$ at which the actual vapor pressure $e$ is equal tot he saturation vapor pressure:  $e=e_s(T_d)$
