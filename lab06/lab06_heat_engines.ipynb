# Part 1: Adiabatic lifting/lowering (3 pts)

First, let's prove our statement from lecture that dry adiabatic lifting followed by dry adiabatic lowering does no work. 

To evaluate different thermodynamic paths, we need functions that generate arrays of state variables ($p$, $\alpha$, $T$, and $\Delta S$) along each process step. 

As an example, you're given a function to output state variables in an adiabatic path.

```python
def adiabatic_path(p_start, T_start, del_s_start, p_end, num_points=50):
    """
    Takes an initial pressure (p_start; Pa), temperature (T_start; K), and Delta entropy (del_s_start, J/kg).

    Lifts/lowers parcel dry adiabatically until p_end, and outputs p (Pa), alpha (m3 kg-1), T (K), Delta s (J/kg) along the way.

    """

    # generate an array ranging from p_start to p_end with num_points evenly spaced values
    p = np.linspace(p_start, p_end, num_points)

    # from Poisson's equation
    T = T_start * (p / p_start) ** kappa

    # from ideal gas law
    alpha = Rd * T / p

    # from first law of thermodynamics in entropy form
    del_s = del_s_start + (cp * np.log(T / T_start)) - (Rd * np.log(p / p_start))

    return (p, alpha, T, del_s)
```

We have a parcel near the surface ($p=1000 hPa$) at $T=20^\circ C$, which we want to lift to $p=800 hPa$. 

Let's plot all our values of entropy relative to the parcel's surface value (so $\Delta s$ represents the increase/decrease in specific entropy relative to this initial state).

```python
p_bottom = 1000 * 100  # Pa
p_top = 800 * 100  # Pa
T_bottom = 20 + 273.15  # K
del_s_bottom = 0  # J /kg

# assign four arrays to hold values of the state variables along this path
# be careful about the order of the function arguments and the output variables!
p_1, alpha_1, T_1, del_s_1 = adiabatic_path(p_bottom, T_bottom, del_s_bottom, p_top)

# to check our output, print pressure values
print(p_1)
```

Let's plot this on a thermodynamic diagram. We'll make a plot with two subpanels: the left side is a $p-\alpha$ diagram and the right side is a $T-s$ diagram.

```python
# set up a figure with two subplots
fig, [ax_palpha, ax_Ts] = plt.subplots(1, 2, figsize=(10, 4))

# plot alpha and p
ax_palpha.plot(alpha_1, p_1, label="Adiabatic Ascent")

# plot T and s
ax_Ts.plot(del_s_1, T_1)

# label our axes
ax_palpha.set_xlabel("Specific Volume (m^3 kg^-1)")
ax_palpha.set_ylabel("Pressure (Pa)")
ax_Ts.set_xlabel("Specific Entropy Change (J kg^-1)")
ax_Ts.set_ylabel("Temperature (K)")

# add legend
ax_palpha.legend()

# set axis limits on entropy plot
ax_Ts.set_xlim(-1, 1)

# show plot
plt.show()
```

# Part 3: Carnot cycle (15 pts)

Many processes in the atmosphere and ocean can be approximated as a Carnot engine (with plenty of caveats). One example is parcel lifting within the planetary boundary layer (PBL).

The PBL is the layer of the atmosphere right above the surface. PBL heights can range from tens of meters (in very cold environments like the poles or at nighttime) to 1-2 kilometers (in very dry environments with a lot of insolation, like deserts). We will start with considering a PBL height up to 840 hPa.

To model parcel motion in the PBL as a Carnot engine, we break it into four steps which we assume happen reversibly:
1. *isothermal expansion*: The surface transfers heat ($Q_{hot}$) into the parcel, allowing the parcel to rise from $p_1$ to $p_2$ and expand isothermally at $T_{hot}$.
2. *adiabatic expansion*: After some time, the parcel becomes buoyant relative to its environment (*more on this in Module 6!*) and starts to ascend dry adiabatically from $p_2$ to $p_3$, exchanging no heat with the surrounding atmosphere ($Q=0$).
3. *isothermal compression*: At the top of the planetary boundary layer, the parcel loses heat ($Q_{cold}$) via outgoing longwave radiation and mixing with cooler air above the boundary layer. The parcel starts to subside slightly from $p_3$ to $p_4$ and is compressed isothermally at $T_{cold}$.
4. *adiabatic compression*: The parcel becomes negatively buoyant relative to its environment and is compressed dry adiabatically ($Q=0$) as it sinks from $p_4$ back to $p_1$. 

```python
# To help build the Carnot cycle, I'll define values for our pressure levels and the surface temperature

# pressures in Pascals
p1 = 100000
p2 = 97500
p3 = 80000
p4 = p3 * (p1 / p2)

T_hot = (
    20 + 273.15
)  # temperature of the isothermal expansion step in K / temperature of the warm reservoir in K


# define the entropy changes relative to the entropy value at the surface
del_s_bottom = 0  # J /kg
```
