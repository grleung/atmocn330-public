# Part 2: Interpreting a Thermodynamic Diagram (5 pts)

Here's some code to plot the three specified paths on a $p-\alpha$ diagram, with some of the labels/comments purposefully missing.

````python
fig, ax = plt.subplots(1, 1)

ax.scatter(alpha1, p1, color="black", label="Initial State")
ax.scatter(alpha2, p2, color="red", label="Final State")

pressure_range = np.linspace(p1, p2, 100)
alpha_range = np.linspace(alpha1, alpha2, 100)

ax.plot(alpha_range, np.full_like(alpha_range, p1), color="purple", ls=":")
ax.plot(np.full_like(pressure_range, alpha2), pressure_range, color="purple", ls=":")

ax.plot(np.full_like(pressure_range, alpha1), pressure_range, color="blue", ls="--")
ax.plot(alpha_range, np.full_like(alpha_range, p2), color="blue", ls="--")


ax.plot(alpha_range, (Rd * T1 / alpha_range) / 100, color="green")

ax.legend(loc="lower right", ncol=2)

ax.set_ylabel("Pressure (hPa)")
ax.set_xlabel("Specific Volume (m^3 kg^-1)")

# set lower axis limit to 0
ax.set_ylim(0)

plt.show()
````
